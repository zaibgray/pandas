What is Pandas
Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language.

[About Pandas](https://pandas.pydata.org/about/index.html)

## The Big Picture
Think of Pandas like Microsoft Excel inside Python:
- DataFrame = The entire Excel Sheet.
- Series = A single column in that sheet.
- Index = The row numbers on the far left.

### 1. Series (1D Column)
A Series is just a single column of data.
- It can hold any data type (numbers, strings, booleans).
- Every item in a Series must be the same type.

### 2. DataFrame (2D Table)
A DataFrame is the main table structure.
- It is made by gluing multiple Series together side-by-side.
- Different columns can have different data types (e.g., text in one column, numbers in another).

### 3. Index (The Labels)
The Index is the list of row labels.
- By default, Pandas automatically numbers rows starting at 0 (0, 1, 2...).
- You can change the index to use words or dates instead of numbers if you want.


**Extracting:** 
If you pull a single column out of a DataFrame, it automatically becomes a Series.

## Pandas Series
A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Series from lists

````python
# strings

names = ["Alice", "Bob", "Charlie", "Diana"]
strings_series = pd.Series(names)
print(strings_series, type(strings_series))
````

````python
# integers

numbers = [10, -5, 42, 0, 7]
numbers_series = pd.Series(numbers)
print(numbers_series, type(numbers_series))
````

### Custom Indexing in Series 

````python
# custome index

name = ["Bob", "Lucky", "Ana"]
id = [1, 4, 7]
name_series = pd.Series(name, index=id)
print(name_series, type(name_series))
````

### Naming to column
In a Pandas Series, the name attribute acts as the column header. 

````python
# setting a name

list_of_food = ["Apple", "Rice", "Bread", "Chicken", "BBQ", "Fish"]
list_index = [1, 3, 6, 7, 2, 4]
list_name = "MyFood"

food_series = pd.Series(list_of_food, index=list_index, name=list_name)
print(food_series, type(food_series))                                           #dtype tells about column data type
````

### Series from Dictionary

When you pass a dictionary to a pd.Series():
- The dictionary Keys become the Index (row labels).
- The dictionary Values become the actual Data.
- Note: You still use the name= parameter separately to give the column a header.

````python
students = {
    101 : "Bob",
    102 : "Alice",
    103 : "Lucy",
    104 : "Ana"
}

dictionary_series = pd.Series(students, name="students_data")
print(dictionary_series)
print(type(dictionary_series))
````
